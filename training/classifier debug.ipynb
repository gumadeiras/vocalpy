```python
import os
import copy
import time
import logging
import numpy             as     np
import matplotlib.pyplot as     plt
from   tqdm              import tqdm
import seaborn           as     sns
import torch
import torch.nn                 as     nn
import torch.optim              as     optim
from   torch.optim              import lr_scheduler as lrs
from   torch.utils.data.sampler import SubsetRandomSampler

import torchvision
from   torchvision       import datasets, models, transforms
```

```python
preds   = np.load('/Users/gustavo/Downloads/preds.npy', allow_pickle=True)
targets = np.load('/Users/gustavo/Downloads/targets.npy', allow_pickle=True)
cm      = np.load('/Users/gustavo/Downloads/confusion.npy', allow_pickle=True)
pcm = np.zeros_like(cm)
classes = ['chevron', 'complex', 'down_fm', 'flat', 'mult_steps', 'rev_chevron', 'short', 'step_down', 'step_up', 'two_steps', 'up_fm']
images  = [532, 64, 188, 132, 7, 10, 204, 112, 185, 392, 239]
for i in range(np.shape(images)[0]):
    pcm[i] = cm[i]/images[i]
```

```python
plt.figure(figsize=(10,10))
ax      = sns.heatmap(cm, annot=True, fmt=".0f")
ax.set_xticklabels(classes, rotation=90)
ax.set_yticklabels(classes, rotation=0)
```

```python
plt.figure(figsize=(10,10))
ax      = sns.heatmap(pcm, annot=True, fmt=".2f")
ax.set_xticklabels(classes, rotation=90)
ax.set_yticklabels(classes, rotation=0)
```

```python
s = []
for i in range(preds.shape[0]):
    s.append(sum(preds[i]==targets[i]))
print(np.sum(s)/np.sum(images))
```
